```python
import os
os.environ.setdefault("PYTORCH_CUDA_ALLOC_CONF", "expandable_segments:True")

import zipfile
import pandas as pd
import numpy as np
from PIL import Image
import torch
from torch.utils.data import Dataset, DataLoader, Subset
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.optim.lr_scheduler import CosineAnnealingLR
from sklearn.metrics import accuracy_score, classification_report, f1_score
from transformers import CLIPModel, CLIPProcessor, CLIPTokenizer
from google.colab import drive

# ==========================================
# 0. Mount Drive, Extract & Filter MVSA-Single
# ==========================================
if not os.path.exists('/content/drive'):
    drive.mount('/content/drive')

zip_file_name = "msva-single.zip"
drive_path = f"/content/drive/MyDrive/{zip_file_name}"
extract_path = "/content/dataset"
os.makedirs(extract_path, exist_ok=True)

if not os.path.exists(os.path.join(extract_path, "labels_clean.csv")):
    if not os.path.exists(drive_path):
        raise FileNotFoundError(
            f"Could not find {drive_path}. Check Google Drive path for {zip_file_name}."
        )
    with zipfile.ZipFile(drive_path, 'r') as zip_ref:
        zip_ref.extractall(extract_path)

img_dir = extract_path
for root, dirs, files in os.walk(extract_path):
    if any(f.endswith(('.jpg', '.png')) for f in files):
        img_dir = root
        break

clean_csv_path = os.path.join(extract_path, 'labels_clean.csv')
if not os.path.exists(clean_csv_path):
    label_file = None
    for root, dirs, files in os.walk(extract_path):
        for file in files:
            if 'labelResultAll' in file and file.endswith('.txt'):
                label_file = os.path.join(root, file)
                break
        if label_file:
            break

    if label_file is None:
        raise FileNotFoundError(f"Could not find labelResultAll*.txt inside {extract_path}.")

    sentiment_map = {'positive': 0, 'neutral': 1, 'negative': 2}
    rows = []
    with open(label_file, 'r', encoding='utf-8', errors='ignore') as f:
        for line in f:
            line = line.strip()
            if not line or line.lower().startswith('id'):
                continue
            parts = line.split('\t') if '\t' in line else line.split(',')
            if len(parts) >= 2:
                sample_id = parts[0].strip()
                pair = [x.strip().lower() for x in parts[1].split(',') if x.strip()]
                if len(pair) >= 2:
                    t_lbl, i_lbl = pair[0], pair[1]
                    if t_lbl in sentiment_map and i_lbl in sentiment_map:
                        if {t_lbl, i_lbl} == {'positive', 'negative'}:
                            continue
                        if t_lbl == i_lbl:
                            final_lbl = sentiment_map[t_lbl]
                        elif t_lbl == 'neutral':
                            final_lbl = sentiment_map[i_lbl]
                        else:
                            final_lbl = sentiment_map[t_lbl]
                        rows.append({'id': sample_id, 'label': final_lbl})

    pd.DataFrame(rows).drop_duplicates(subset=['id']).to_csv(clean_csv_path, index=False)

# ==========================================
# 1. Dataset & Cross-Modal Attention
# ==========================================
class MVSACLIPDataset(Dataset):
    def __init__(self, csv_file, img_dir, processor):
        self.df = pd.read_csv(csv_file)
        self.img_dir = img_dir
        self.processor = processor

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        sample_id = str(row['id'])
        label = int(row['label'])

        img_path_jpg = os.path.join(self.img_dir, f"{sample_id}.jpg")
        img_path_png = os.path.join(self.img_dir, f"{sample_id}.png")
        if os.path.exists(img_path_jpg):
            image = Image.open(img_path_jpg).convert("RGB")
        elif os.path.exists(img_path_png):
            image = Image.open(img_path_png).convert("RGB")
        else:
            image = Image.new("RGB", (224, 224), color="gray")

        txt_path = os.path.join(self.img_dir, f"{sample_id}.txt")
        if os.path.exists(txt_path):
            with open(txt_path, "r", encoding="utf-8", errors="ignore") as tf:
                text = tf.read().strip()
        else:
            text = ""

        enc = self.processor(
            text=[text],
            images=image,
            padding="max_length",
            max_length=77,
            truncation=True,
            return_tensors="pt"
        )
        return {
            "input_ids": enc["input_ids"].squeeze(0),
            "attention_mask": enc["attention_mask"].squeeze(0),
            "pixel_values": enc["pixel_values"].squeeze(0),
            "label": torch.tensor(label, dtype=torch.long)
        }


class CrossModalAttention(nn.Module):
    def __init__(self, embed_dim):
        super().__init__()
        self.scale = 1.0 / (embed_dim ** 0.5)

    def forward(self, q, k, v):
        scores = torch.matmul(q, k.transpose(-2, -1)) * self.scale
        attn = F.softmax(scores, dim=-1)
        return torch.matmul(attn, v)


def masked_mean_pool(features, mask):
    mask_expanded = mask.unsqueeze(-1).float()
    summed = (features * mask_expanded).sum(dim=1)
    counts = mask_expanded.sum(dim=1).clamp(min=1.0)
    return summed / counts


def _extract_tensor_output(output):
    if torch.is_tensor(output):
        return output
    for attr in ("text_embeds", "image_embeds", "pooler_output", "last_hidden_state"):
        val = getattr(output, attr, None)
        if torch.is_tensor(val):
            return val[:, 0] if (attr == "last_hidden_state" and val.ndim == 3) else val
    if isinstance(output, (tuple, list)) and len(output) > 0:
        first = output[0]
        if torch.is_tensor(first):
            return first[:, 0] if first.ndim == 3 else first
    raise TypeError(f"Unsupported model output type for normalization: {type(output)}")


# ==========================================
# 2. MAMSA (CLIP ViT-B/16)
# ==========================================
class MAMSA_CLIP(nn.Module):
    def __init__(self, clip_name="openai/clip-vit-base-patch16", num_classes=3, embed_dim=512, dropout=0.2):
        super().__init__()
        self.clip = CLIPModel.from_pretrained(clip_name)
        self.embed_dim = embed_dim

        with torch.no_grad():
            tok = CLIPTokenizer.from_pretrained(clip_name)
            prompts = ["a positive sentiment", "a neutral sentiment", "a negative sentiment"]
            inputs = tok(prompts, padding=True, return_tensors="pt")
            raw_emb = self.clip.get_text_features(**inputs)
            label_emb = _extract_tensor_output(raw_emb).float()
            if label_emb.size(-1) != embed_dim:
                label_emb = self.clip.text_projection(label_emb)
            label_emb = F.normalize(label_emb, dim=-1)
        self.sentiment_matrix = nn.Parameter(label_emb * 0.5)

        self.w_c = nn.Linear(embed_dim * 2, 1)
        self.ca_text = CrossModalAttention(embed_dim)
        self.ca_image = CrossModalAttention(embed_dim)
        self.sta = CrossModalAttention(embed_dim)
        self.sia = CrossModalAttention(embed_dim)

        self.hf_proj_c = nn.Linear(embed_dim * 2, embed_dim)
        self.hf_attn1 = CrossModalAttention(embed_dim)
        self.hf_layer1_linear = nn.Linear(embed_dim, embed_dim)
        self.hf_attn2 = CrossModalAttention(embed_dim)
        self.dropout = nn.Dropout(dropout)
        self.classifier = nn.Linear(embed_dim, num_classes)

    def forward(self, input_ids, attention_mask, pixel_values):
        batch_size = input_ids.size(0)

        text_feat = self.clip.text_model(input_ids=input_ids, attention_mask=attention_mask).last_hidden_state
        vision_feat = self.clip.vision_model(pixel_values=pixel_values).last_hidden_state

        if text_feat.size(-1) != self.embed_dim:
            text_feat = self.clip.text_projection(text_feat)
        if vision_feat.size(-1) != self.embed_dim:
            vision_feat = self.clip.visual_projection(vision_feat)

        # 1. AAI
        pooled_text = masked_mean_pool(text_feat, attention_mask)
        pooled_image = vision_feat.mean(dim=1)
        hat_c = torch.cat([pooled_text, pooled_image], dim=1)

        theta = torch.sigmoid(self.w_c(hat_c)).unsqueeze(-1)
        text_prime = theta * text_feat
        image_prime = (1.0 - theta) * vision_feat

        tilde_text = self.ca_text(q=image_prime, k=text_feat, v=text_feat)
        tilde_image = self.ca_image(q=text_prime, k=vision_feat, v=vision_feat)

        X_C = torch.cat([tilde_text.mean(dim=1), masked_mean_pool(tilde_image, attention_mask)], dim=1)
        X_C = self.hf_proj_c(X_C)

        # 2. SA
        Ms = self.sentiment_matrix.unsqueeze(0).expand(batch_size, -1, -1)
        X_ST = self.sta(q=Ms, k=text_feat, v=text_feat).mean(dim=1)
        X_SI = self.sia(q=Ms, k=vision_feat, v=vision_feat).mean(dim=1)

        # 3. Hierarchical Fusion
        X_S = self.hf_layer1_linear(self.hf_attn1(q=X_ST.unsqueeze(1), k=X_SI.unsqueeze(1), v=X_SI.unsqueeze(1)).squeeze(1))
        X_final = self.hf_attn2(q=X_S.unsqueeze(1), k=X_C.unsqueeze(1), v=X_C.unsqueeze(1)).squeeze(1)

        return self.classifier(self.dropout(X_final))


# ==========================================
# 3. Split, Oversample Train & Train CLIP MAMSA
# ==========================================
if torch.cuda.is_available():
    torch.cuda.empty_cache()

clip_name = "openai/clip-vit-base-patch16"
processor = CLIPProcessor.from_pretrained(clip_name)
full_dataset = MVSACLIPDataset(clean_csv_path, img_dir, processor)

n = len(full_dataset)
n_train, n_val = int(0.8 * n), int(0.1 * n)
n_test = n - n_train - n_val
train_ds, val_ds, test_ds = torch.utils.data.random_split(
    full_dataset, [n_train, n_val, n_test], generator=torch.Generator().manual_seed(42)
)

train_idx = train_ds.indices
train_lbls = np.array([full_dataset.df.iloc[i]['label'] for i in train_idx])
max_cnt = np.max(np.bincount(train_lbls))
balanced_idx = list(train_idx)
for c in range(3):
    locs = np.where(train_lbls == c)[0]
    if 0 < len(locs) < max_cnt:
        extra = np.random.choice(locs, size=max_cnt - len(locs), replace=True)
        balanced_idx.extend([train_idx[l] for l in extra])

batch_size = 16
grad_accum_steps = 2

train_loader = DataLoader(Subset(full_dataset, balanced_idx), batch_size=batch_size, shuffle=True, num_workers=2, pin_memory=True)
val_loader = DataLoader(val_ds, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True)
test_loader = DataLoader(test_ds, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = MAMSA_CLIP(clip_name=clip_name).to(device)

backbone_ids = {id(p) for p in model.clip.parameters()}
optimizer = optim.AdamW([
    {'params': [p for p in model.parameters() if id(p) in backbone_ids], 'lr': 1e-5},
    {'params': [p for p in model.parameters() if id(p) not in backbone_ids], 'lr': 2e-4}
], weight_decay=1e-4)

epochs = 12
scheduler = CosineAnnealingLR(optimizer, T_max=epochs, eta_min=1e-6)
criterion = nn.CrossEntropyLoss()
scaler = torch.amp.GradScaler('cuda', enabled=(device.type == 'cuda'))

os.makedirs("saved_models", exist_ok=True)
ckpt_path = "saved_models/mamsa_clip_best.pth"
best_score = -1.0

for epoch in range(epochs):
    model.train()
    total_loss = 0.0
    optimizer.zero_grad()

    for step, b in enumerate(train_loader):
        with torch.amp.autocast('cuda', enabled=(device.type == 'cuda')):
            out = model(b["input_ids"].to(device), b["attention_mask"].to(device), b["pixel_values"].to(device))
            loss = criterion(out, b["label"].to(device)) / grad_accum_steps

        scaler.scale(loss).backward()
        total_loss += loss.item() * grad_accum_steps

        if (step + 1) % grad_accum_steps == 0 or (step + 1) == len(train_loader):
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            scaler.step(optimizer)
            scaler.update()
            optimizer.zero_grad()

    scheduler.step()

    model.eval()
    v_preds, v_trues = [], []
    with torch.no_grad():
        for b in val_loader:
            with torch.amp.autocast('cuda', enabled=(device.type == 'cuda')):
                out = model(b["input_ids"].to(device), b["attention_mask"].to(device), b["pixel_values"].to(device))
            v_preds.extend(torch.argmax(out, dim=1).cpu().numpy())
            v_trues.extend(b["label"].numpy())

    v_acc = accuracy_score(v_trues, v_preds)
    v_f1 = f1_score(v_trues, v_preds, average='macro', zero_division=0)
    score = 0.5 * v_acc + 0.5 * v_f1
    print(f"Epoch [{epoch+1}/{epochs}] | Loss: {total_loss/len(train_loader):.4f} | Val Acc: {v_acc*100:.2f}% | Val Macro F1: {v_f1*100:.2f}%")
    if score > best_score:
        best_score = score
        torch.save(model.state_dict(), ckpt_path)

model.load_state_dict(torch.load(ckpt_path, map_location=device))
model.eval()
t_preds, t_trues = [], []
with torch.no_grad():
    for b in test_loader:
        with torch.amp.autocast('cuda', enabled=(device.type == 'cuda')):
            out = model(b["input_ids"].to(device), b["attention_mask"].to(device), b["pixel_values"].to(device))
        t_preds.extend(torch.argmax(out, dim=1).cpu().numpy())
        t_trues.extend(b["label"].numpy())

print(f"\n[CLIP MAMSA] Test Acc: {accuracy_score(t_trues, t_preds)*100:.2f}% | Macro F1: {f1_score(t_trues, t_preds, average='macro', zero_division=0)*100:.2f}%")
print(classification_report(t_trues, t_preds, target_names=["Positive", "Neutral", "Negative"], zero_division=0))
```

```python
import os
os.environ.setdefault("PYTORCH_CUDA_ALLOC_CONF", "expandable_segments:True")

import gc
import pandas as pd
import numpy as np
from PIL import Image
import torch
from torch.utils.data import Dataset, DataLoader, Subset
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.optim.lr_scheduler import CosineAnnealingLR
from sklearn.metrics import accuracy_score, classification_report, f1_score
from transformers import AutoTokenizer, AutoModel, CLIPModel, CLIPProcessor

# Free Cell 1 GPU memory before loading BERT + CLIP Vision
for var_name in ("model", "optimizer", "scaler", "scheduler", "train_loader", "val_loader", "test_loader"):
    if var_name in globals():
        del globals()[var_name]
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()

clean_csv_path = "/content/dataset/labels_clean.csv"
img_dir = "/content/dataset"
for root, dirs, files in os.walk("/content/dataset"):
    if any(f.endswith(('.jpg', '.png')) for f in files):
        img_dir = root
        break

class MVSABERTDataset(Dataset):
    def __init__(self, csv_file, img_dir, tokenizer, clip_processor, max_len=96):
        self.df = pd.read_csv(csv_file)
        self.img_dir = img_dir
        self.tokenizer = tokenizer
        self.clip_processor = clip_processor
        self.max_len = max_len

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        sample_id = str(row['id'])
        label = int(row['label'])

        img_path_jpg = os.path.join(self.img_dir, f"{sample_id}.jpg")
        img_path_png = os.path.join(self.img_dir, f"{sample_id}.png")
        if os.path.exists(img_path_jpg):
            image = Image.open(img_path_jpg).convert("RGB")
        elif os.path.exists(img_path_png):
            image = Image.open(img_path_png).convert("RGB")
        else:
            image = Image.new("RGB", (224, 224), color="gray")

        txt_path = os.path.join(self.img_dir, f"{sample_id}.txt")
        if os.path.exists(txt_path):
            with open(txt_path, "r", encoding="utf-8", errors="ignore") as tf:
                text = tf.read().strip()
        else:
            text = ""

        t_enc = self.tokenizer(
            text,
            padding="max_length",
            max_length=self.max_len,
            truncation=True,
            return_tensors="pt"
        )
        v_enc = self.clip_processor(images=image, return_tensors="pt")

        return {
            "input_ids": t_enc["input_ids"].squeeze(0),
            "attention_mask": t_enc["attention_mask"].squeeze(0),
            "pixel_values": v_enc["pixel_values"].squeeze(0),
            "label": torch.tensor(label, dtype=torch.long)
        }


class CrossModalAttention(nn.Module):
    def __init__(self, embed_dim):
        super().__init__()
        self.scale = 1.0 / (embed_dim ** 0.5)

    def forward(self, q, k, v):
        scores = torch.matmul(q, k.transpose(-2, -1)) * self.scale
        attn = F.softmax(scores, dim=-1)
        return torch.matmul(attn, v)


def masked_mean_pool(features, mask):
    mask_expanded = mask.unsqueeze(-1).float()
    return (features * mask_expanded).sum(dim=1) / mask_expanded.sum(dim=1).clamp(min=1.0)


class MAMSA_BERT(nn.Module):
    def __init__(self, bert_name="bert-base-uncased", clip_vis_name="openai/clip-vit-base-patch16",
                 num_classes=3, embed_dim=512, dropout=0.25):
        super().__init__()
        self.bert = AutoModel.from_pretrained(bert_name)
        self.clip_vision = CLIPModel.from_pretrained(clip_vis_name).vision_model

        self.text_proj = nn.Sequential(
            nn.Linear(self.bert.config.hidden_size, embed_dim),
            nn.LayerNorm(embed_dim)
        )
        self.vis_proj = nn.Sequential(
            nn.Linear(self.clip_vision.config.hidden_size, embed_dim),
            nn.LayerNorm(embed_dim)
        )

        with torch.no_grad():
            tok = AutoTokenizer.from_pretrained(bert_name)
            lbl_texts = ["positive sentiment", "neutral sentiment", "negative sentiment"]
            inp = tok(lbl_texts, padding=True, truncation=True, return_tensors="pt")
            out = self.bert(**inp).last_hidden_state[:, 0, :]
            init_ms = self.text_proj(out)
        self.sentiment_matrix = nn.Parameter(init_ms)

        self.w_c = nn.Linear(embed_dim * 2, 1)
        self.ca_text = CrossModalAttention(embed_dim)
        self.ca_image = CrossModalAttention(embed_dim)
        self.sta = CrossModalAttention(embed_dim)
        self.sia = CrossModalAttention(embed_dim)

        self.hf_proj_c = nn.Linear(embed_dim * 2, embed_dim)
        self.hf_attn1 = CrossModalAttention(embed_dim)
        self.hf_layer1_linear = nn.Linear(embed_dim, embed_dim)
        self.hf_attn2 = CrossModalAttention(embed_dim)
        self.dropout = nn.Dropout(dropout)
        self.classifier = nn.Linear(embed_dim, num_classes)

    def forward(self, input_ids, attention_mask, pixel_values):
        batch_size = input_ids.size(0)

        text_feat = self.text_proj(self.bert(input_ids=input_ids, attention_mask=attention_mask).last_hidden_state)
        vision_feat = self.vis_proj(self.clip_vision(pixel_values=pixel_values).last_hidden_state)

        # 1. AAI
        pooled_text = masked_mean_pool(text_feat, attention_mask)
        pooled_image = vision_feat.mean(dim=1)
        hat_c = torch.cat([pooled_text, pooled_image], dim=1)

        theta = torch.sigmoid(self.w_c(hat_c)).unsqueeze(-1)
        text_prime = theta * text_feat
        image_prime = (1.0 - theta) * vision_feat

        tilde_text = self.ca_text(q=image_prime, k=text_feat, v=text_feat)
        tilde_image = self.ca_image(q=text_prime, k=vision_feat, v=vision_feat)

        X_C = torch.cat([tilde_text.mean(dim=1), masked_mean_pool(tilde_image, attention_mask)], dim=1)
        X_C = self.hf_proj_c(X_C)

        # 2. SA
        Ms = self.sentiment_matrix.unsqueeze(0).expand(batch_size, -1, -1)
        X_ST = self.sta(q=Ms, k=text_feat, v=text_feat).mean(dim=1)
        X_SI = self.sia(q=Ms, k=vision_feat, v=vision_feat).mean(dim=1)

        # 3. Hierarchical Fusion
        X_S = self.hf_layer1_linear(self.hf_attn1(q=X_ST.unsqueeze(1), k=X_SI.unsqueeze(1), v=X_SI.unsqueeze(1)).squeeze(1))
        X_final = self.hf_attn2(q=X_S.unsqueeze(1), k=X_C.unsqueeze(1), v=X_C.unsqueeze(1)).squeeze(1)

        return self.classifier(self.dropout(X_final))


tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
clip_proc = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch16")
bert_dataset = MVSABERTDataset(clean_csv_path, img_dir, tokenizer, clip_proc)

n = len(bert_dataset)
n_train, n_val = int(0.8 * n), int(0.1 * n)
n_test = n - n_train - n_val
b_train_ds, b_val_ds, b_test_ds = torch.utils.data.random_split(
    bert_dataset, [n_train, n_val, n_test], generator=torch.Generator().manual_seed(42)
)

t_idx = b_train_ds.indices
t_lbls = np.array([bert_dataset.df.iloc[i]['label'] for i in t_idx])
max_c = np.max(np.bincount(t_lbls))
bal_idx = list(t_idx)
for c in range(3):
    locs = np.where(t_lbls == c)[0]
    if 0 < len(locs) < max_c:
        extra = np.random.choice(locs, size=max_c - len(locs), replace=True)
        bal_idx.extend([t_idx[l] for l in extra])

batch_size = 16
grad_accum_steps = 2

b_train_loader = DataLoader(Subset(bert_dataset, bal_idx), batch_size=batch_size, shuffle=True, num_workers=2, pin_memory=True)
b_val_loader = DataLoader(b_val_ds, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True)
b_test_loader = DataLoader(b_test_ds, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
bert_model = MAMSA_BERT().to(device)

bb_ids = {id(p) for p in list(bert_model.bert.parameters()) + list(bert_model.clip_vision.parameters())}
optimizer = optim.AdamW([
    {'params': [p for p in bert_model.parameters() if id(p) in bb_ids], 'lr': 1.5e-5},
    {'params': [p for p in bert_model.parameters() if id(p) not in bb_ids], 'lr': 2e-4}
], weight_decay=1e-4)

epochs = 12
scheduler = CosineAnnealingLR(optimizer, T_max=epochs, eta_min=1e-6)
criterion = nn.CrossEntropyLoss()
scaler = torch.amp.GradScaler('cuda', enabled=(device.type == 'cuda'))

os.makedirs("saved_models", exist_ok=True)
b_ckpt = "saved_models/mamsa_bert_best.pth"
best_b_score = -1.0

for epoch in range(epochs):
    bert_model.train()
    tr_loss = 0.0
    optimizer.zero_grad()

    for step, b in enumerate(b_train_loader):
        with torch.amp.autocast('cuda', enabled=(device.type == 'cuda')):
            out = bert_model(b["input_ids"].to(device), b["attention_mask"].to(device), b["pixel_values"].to(device))
            loss = criterion(out, b["label"].to(device)) / grad_accum_steps

        scaler.scale(loss).backward()
        total_loss_step = loss.item() * grad_accum_steps
        tr_loss += total_loss_step

        if (step + 1) % grad_accum_steps == 0 or (step + 1) == len(b_train_loader):
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(bert_model.parameters(), 1.0)
            scaler.step(optimizer)
            scaler.update()
            optimizer.zero_grad()

    scheduler.step()

    bert_model.eval()
    vp, vt = [], []
    with torch.no_grad():
        for b in b_val_loader:
            with torch.amp.autocast('cuda', enabled=(device.type == 'cuda')):
                out = bert_model(b["input_ids"].to(device), b["attention_mask"].to(device), b["pixel_values"].to(device))
            vp.extend(torch.argmax(out, dim=1).cpu().numpy())
            vt.extend(b["label"].numpy())

    va = accuracy_score(vt, vp)
    vf = f1_score(vt, vp, average='macro', zero_division=0)
    sc = 0.5 * va + 0.5 * vf
    print(f"[BERT] Epoch [{epoch+1}/{epochs}] | Loss: {tr_loss/len(b_train_loader):.4f} | Val Acc: {va*100:.2f}% | Val Macro F1: {vf*100:.2f}%")
    if sc > best_b_score:
        best_b_score = sc
        torch.save(bert_model.state_dict(), b_ckpt)

bert_model.load_state_dict(torch.load(b_ckpt, map_location=device))
bert_model.eval()
tp, tt = [], []
with torch.no_grad():
    for b in b_test_loader:
        with torch.amp.autocast('cuda', enabled=(device.type == 'cuda')):
            out = bert_model(b["input_ids"].to(device), b["attention_mask"].to(device), b["pixel_values"].to(device))
        tp.extend(torch.argmax(out, dim=1).cpu().numpy())
        tt.extend(b["label"].numpy())

print(f"\n[BERT+CLIP MAMSA] Test Acc: {accuracy_score(tt, tp)*100:.2f}% | Macro F1: {f1_score(tt, tp, average='macro', zero_division=0)*100:.2f}%")
print(classification_report(tt, tp, target_names=["Positive", "Neutral", "Negative"], zero_division=0))
```
